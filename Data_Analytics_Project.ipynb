#Phase - 4 (Create New Columns)

```python
# creat age_group column  with age below 25= young,25-35=mid,above 40= senior.
df["Age_Group"] = df["Age"].apply( lambda x : "Young" if x<25 else "Mid" if x<=35 else "Senior" )
```

```python
# create Performance_Level with 90+ score = excellent, 80-89=good, below 80=needs improvement.
df["Performance_Level"] = df["Performance_Score"].apply( lambda x: "Excellent" if x>=90 else "Good" if x>=80 else "Needs Improvement")
```

#Phase-6 (Matplotlib Visualizations)

```python
#importing
import matplotlib.pyplot as plt
```

```python
#Bar Chart — Average Salary by Department

plt.bar(
    df.groupby("Department")["Salary"].mean().index,
    df.groupby("Department")["Salary"].mean().values
)

plt.title("Average Salary by Department")
plt.xlabel("Department")
plt.ylabel("Average Salary")
plt.show()
```

```python
#Line Chart — Average Performance by Department
plt.plot(
    df.groupby("Department")["Performance_Score"].mean(),
    marker="o"
)
plt.title("Average Performance_Score by Department")
plt.xlabel("Department")
plt.ylabel("Average Performance Score")
plt.show()
```

```python
#Pie Chart — Employees by Department
df["Department"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%"
)

plt.title("Employees by Department")
plt.ylabel("")

plt.show()
```

```python
#Histogram — Salary Distribution.
plt.hist(df["Salary"],bins=5)

plt.title("Salary Distribution")
plt.xlabel("Number of Employees")
plt.ylabel("Salary of Employees")

plt.grid()
plt.show()
```

#Phase-7 (Seaborn Visualizations)

```python
#importing seaborn
import seaborn as sns
```

```python
#Countplot — Employees by Department
sns.countplot(
    x="Department",
    data=df
)
plt.xlabel("Department")
plt.ylabel("Number of Employees")
plt.title("Employees by Department")

plt.show()
```

```python
#Barplot — Average Salary by Department.
sns.barplot(
    x="Department",
    y="Salary",
    data=df
)

plt.title("Average Salary by Departments")
plt.grid()
plt.show()
```

```python
#Boxplot — Salary Distribution by Department.
sns.boxplot(
    x="Department",
    y="Salary",
    data=df
)

plt.xlabel("Department")
plt.ylabel("Salary")
plt.title("Salary Distribution by Department")
plt.grid()
plt.legend()
plt.show()
```

```python
#Heatmap — Correlation Between Numerical Columns.
correlation = df.corr(numeric_only=True)
sns.heatmap(correlation,annot=True)
plt.title("Correlation Heatmap")
plt.show()
```
